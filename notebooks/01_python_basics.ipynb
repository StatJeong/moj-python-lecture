## 1.3 자료형

## 1.4 연산자

### 산술 연산자

| 연산자 | 뜻 | 예 | 결과 |
|---|---|---|---|
| `+` | 더하기 | `30 + 6` | `36` |
| `-` | 빼기 | `30 - 6` | `24` |
| `*` | 곱하기 | `30 * 2` | `60` |
| `/` | 나누기 | `30 / 12` | `2.5` |
| `//` | 몫 | `30 // 12` | `2` |
| `%` | 나머지 | `30 % 12` | `6` |
| `**` | 거듭제곱 | `2 ** 3` | `8` |

몫(`//`)과 나머지(`%`)를 쓰면 형기 30개월을 "2년 6개월"로 바꿀 수 있습니다.

```python
형기_개월 = 30
년 = 형기_개월 // 12
개월 = 형기_개월 % 12
print(년, "년", 개월, "개월")
```

2025년 전국 교정시설의 수용 정원은 50,614명, 1일 평균 수용인원은 63,680명이었습니다(2026 교정통계연보). 수용률을 계산해 봅니다.

```python
정원 = 50614
현원 = 63680
수용률 = 현원 / 정원 * 100
print(수용률)
print(round(수용률, 1))
```

곱셈과 나눗셈이 덧셈과 뺄셈보다 먼저 계산됩니다. 먼저 계산할 부분은 괄호로 묶습니다.

```python
print(현원 - 정원)                 # 정원을 넘는 인원
print((현원 - 정원) / 정원 * 100)  # 정원 대비 초과 비율(%)
```

### 비교 연산자

비교 연산의 결과는 `True` 또는 `False`입니다.

| 연산자 | 뜻 |
|---|---|
| `==` | 같다 |
| `!=` | 다르다 |
| `>`, `>=` | 크다, 크거나 같다 |
| `<`, `<=` | 작다, 작거나 같다 |

```python
나이 = 24
죄명 = "절도"
print(나이 < 30)
print(나이 == 30)
print(죄명 == "사기")
print(죄명 != "사기")
```

`=`는 저장, `==`는 같은지 비교입니다. 처음에 가장 많이 헷갈리는 부분입니다.

### 논리 연산자

| 연산자 | 뜻 |
|---|---|
| `and` | 둘 다 참이면 참 |
| `or` | 하나라도 참이면 참 |
| `not` | 참과 거짓을 뒤집음 |

```python
나이 = 24
징벌횟수 = 2
직업훈련 = False

print(나이 < 30 and 징벌횟수 > 0)
print(나이 < 30 or 징벌횟수 > 3)
print(not 직업훈련)
```

## 1.6 리스트

여러 값을 순서대로 묶을 때는 리스트를 씁니다. 대괄호 `[ ]` 안에 쉼표로 구분해 넣습니다.

```python
나이목록 = [24, 31, 45, 28, 52]
print(나이목록)
```

## 1.8 조건문

조건에 따라 다른 코드를 실행할 때 `if`를 씁니다.

```python
나이 = 24

if 나이 < 30:
    print("30세 미만")
elif 나이 < 40:
    print("30대")
else:
    print("40세 이상")
```

- 조건 끝에 콜론(`:`)을 붙입니다.
- 조건이 참일 때 실행할 코드는 4칸 들여 씁니다. 파이썬은 들여쓰기로 어디까지가 `if` 안인지 구분합니다.
- `elif`는 "앞 조건이 거짓이고 이 조건이 참이면", `else`는 "앞의 조건이 모두 거짓이면"이라는 뜻입니다.

```python
징벌횟수 = 3

if 징벌횟수 >= 3:
    print("상담 우선 검토")
else:
    print("일반")
```

## 1.9 반복문

리스트의 값을 하나씩 꺼내 같은 작업을 반복할 때 `for`를 씁니다. 형기를 하나씩 더해 합계를 구해 봅니다.

```python
형기목록 = [24, 36, 24, 30]
합계 = 0
for 형기 in 형기목록:
    합계 = 합계 + 형기
print("형기 합계:", 합계)
```

아래 움직이는 그림은 위 코드가 한 줄씩 실행되는 과정입니다. 노란 줄이 방금 실행한 줄이고, 오른쪽에 그 결과로 바뀐 변수 값이 초록색으로 표시됩니다.

![for 반복문이 한 줄씩 실행되는 과정](https://raw.githubusercontent.com/StatJeong/moj-python-lecture/main/images/loop_sum.gif)

반복할 때마다 `형기`와 `합계`가 바뀌는 과정을 표로 정리하면 다음과 같습니다.

| 반복 | 꺼낸 값 `형기` | 실행 전 `합계` | 실행 후 `합계` |
|---|---|---|---|
| 1회째 | 24 | 0 | 24 |
| 2회째 | 36 | 24 | 60 |
| 3회째 | 24 | 60 | 84 |
| 4회째 | 30 | 84 | 114 |

`for` 안에 `if`를 넣으면 조건에 맞는 값만 처리할 수 있습니다.

```python
나이목록 = [24, 31, 45, 28]
인원 = 0
for 나이 in 나이목록:
    if 나이 < 30:
        인원 = 인원 + 1
        print(나이, "→ 30세 미만")
print("30세 미만 인원:", 인원)
```

![for 안의 if가 참과 거짓에 따라 다르게 실행되는 과정](https://raw.githubusercontent.com/StatJeong/moj-python-lecture/main/images/loop_if.gif)

정해진 횟수만큼 반복할 때는 `range()`를 씁니다. `range(3)`은 0, 1, 2를 차례로 만듭니다.

```python
for i in range(3):
    print(i, "번째 반복")
```

딕셔너리는 `.items()`로 이름표와 값을 함께 꺼낼 수 있습니다.

```python
징벌 = {"1001": 2, "1002": 0, "1004": 3}
for 번호, 횟수 in 징벌.items():
    print(번호, "번 징벌", 횟수, "회")
```

## 1.10 함수

자주 하는 계산을 함수로 만들어 두면, 값만 바꿔 넣어 같은 계산을 몇 번이고 다시 할 수 있습니다. 지금까지 쓴 `print()`, `len()`, `sum()`, `round()`, `type()`도 파이썬에 미리 만들어져 있는 함수입니다. 여기서는 함수를 직접 만듭니다.

### 함수 만들기와 부르기

```python
def 수용률계산(현원, 정원):
    비율 = 현원 / 정원 * 100
    return round(비율, 1)

A = 수용률계산(1523, 1200)
B = 수용률계산(1080, 900)
print("A교도소:", A, "/ B교도소:", B)
```

| 부분 | 뜻 |
|---|---|
| `def 수용률계산(현원, 정원):` | `수용률계산`이라는 함수를 만든다. 괄호 안은 함수에 넣을 값의 이름(매개변수) |
| 들여 쓴 줄 | 함수를 부를 때 실행되는 내용 |
| `return` | 결과를 돌려주고 함수를 끝낸다 |
| `수용률계산(1523, 1200)` | 함수를 부른다. 1523은 `현원`에, 1200은 `정원`에 들어간다 |

아래 그림에서 함수를 부르면 값이 함수 안으로 들어가고(점선 상자), `return`으로 결과를 돌려준 뒤 함수 안의 변수가 사라지는 과정을 볼 수 있습니다. `def` 줄을 실행할 때는 함수를 만들어 두기만 하고, 안쪽 줄은 함수를 부를 때 실행됩니다.

![함수를 부르고 결과를 돌려받는 과정](https://raw.githubusercontent.com/StatJeong/moj-python-lecture/main/images/func_call.gif)

### 기본값과 이름 붙여 넣기

매개변수에 기본값을 정해 두면, 값을 넣지 않았을 때 기본값을 씁니다. 값을 넣을 때 `단위="개월"`처럼 이름을 붙여 넣을 수도 있습니다.

```python
def 형기표시(개월, 단위="년"):
    if 단위 == "년":
        return f"{개월 // 12}년 {개월 % 12}개월"
    return f"{개월}개월"

print(형기표시(30))
print(형기표시(30, 단위="개월"))
print(형기표시(개월=42))
```

### 여러 값 돌려주기

`return` 뒤에 쉼표로 여러 값을 쓰면 한 번에 돌려받을 수 있습니다.

```python
def 형기나누기(개월):
    return 개월 // 12, 개월 % 12

년, 개월 = 형기나누기(30)
print(년, "년", 개월, "개월")
```

### 함수 안의 조건문

함수 안에서 조건에 따라 다른 값을 돌려줄 수 있습니다. `return`을 만나면 그 자리에서 함수가 끝나므로 아래 줄은 실행되지 않습니다.

```python
def 연령대(나이):
    if 나이 < 30:
        return "30세 미만"
    elif 나이 < 40:
        return "30대"
    return "40세 이상"

for 나이 in [24, 45]:
    print(나이, 연령대(나이))
```

![함수 안의 조건문이 실행되는 과정](https://raw.githubusercontent.com/StatJeong/moj-python-lecture/main/images/func_if.gif)

함수 밖의 `나이`와 함수 안의 `나이`는 이름만 같을 뿐 서로 다른 변수입니다. 그림에서 두 값이 따로 표시되는 것을 확인할 수 있습니다.

### 리스트를 받아 결과를 뽑는 함수

반복문과 조건문을 함수 안에 넣으면, 기준만 바꿔 원하는 결과를 뽑을 수 있습니다.

```python
def 기준이상인원(값목록, 기준):
    인원 = 0
    for 값 in 값목록:
        if 값 >= 기준:
            인원 = 인원 + 1
    return 인원

징벌목록 = [2, 0, 0, 3, 0, 1, 1, 1, 2, 0]
print("징벌 1회 이상:", 기준이상인원(징벌목록, 1), "명")
print("징벌 2회 이상:", 기준이상인원(징벌목록, 2), "명")
```

2장에서는 같은 방식으로 표 전체를 받아 원하는 결과를 돌려주는 함수를 만듭니다.
