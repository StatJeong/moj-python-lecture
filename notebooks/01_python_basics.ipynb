## 1.2 변수

변수는 값에 이름을 붙여 저장해 두는 것입니다.

````python
수용번호 = 1001
나이 = 24
죄명 = "절도"
print(수용번호, 나이, 죄명)
````

- `=`는 "같다"는 뜻이 아니라 오른쪽 값을 왼쪽 이름에 저장한다는 뜻입니다.
- 변수 이름에는 한글을 쓸 수 있습니다. 숫자로 시작할 수 없고 띄어쓰기도 안 되므로, 띄어 쓰고 싶으면 밑줄을 씁니다(`형기_개월`).
- 이미 있는 변수에 다시 저장하면 값이 바뀝니다.

> [주의] 변수 이름 규칙
>
> | 쓸 수 없는 이름 | 이유 | 고친 예 |
> |---|---|---|
> | `1번수용자` | 숫자로 시작 | `수용자1` |
> | `수용 번호` | 띄어쓰기 | `수용_번호` |
> | `형기-개월` | `-`를 빼기 기호로 읽음 | `형기_개월` |
> | `if`, `for`, `class`, `True` | 파이썬이 문법에 쓰는 단어(예약어) | `조건`, `구분` |
>
> 규칙을 어기면 실행 전에 `SyntaxError`가 납니다. 예약어 목록은 `import keyword` 다음 `print(keyword.kwlist)`로 볼 수 있습니다.


> [참고] 대문자와 소문자는 다른 이름
>
> `Name`과 `name`은 서로 다른 변수입니다. `Name = 1`로 만들고 `print(name)`을 실행하면 `NameError: name 'name' is not defined. Did you mean: 'Name'?`이 납니다. 오류 메시지의 `Did you mean`에 비슷한 이름이 나오면 오타를 의심합니다.


> [주의] sum, max, list 같은 이름을 변수로 쓰지 않기
>
> `sum`, `max`, `min`, `len`, `list`, `print`, `type`, `round`는 파이썬에 이미 있는 함수 이름입니다. 이 이름에 값을 저장하면 그 뒤로 함수가 동작하지 않습니다.
>
> ```python
> sum = 10
> sum([1, 2, 3])   # TypeError: 'int' object is not callable
> ```
>
> 이미 덮어썼다면 `del sum`으로 지우거나 세션을 다시 시작합니다. 합계는 `합계`, `total`처럼 다른 이름에 저장합니다.

````python
나이 = 나이 + 1
print(나이)
````

## 1.3 자료형

값에는 종류가 있습니다. 데이터를 다룰 때 가장 많이 쓰는 네 가지는 다음과 같습니다.

| 자료형 | 이름 | 예 | 교정 데이터에서 |
|---|---|---|---|
| `int` | 정수 | `24`, `1001` | 나이, 형기(개월), 수용번호 |
| `float` | 실수 | `126.9` | 수용률 |
| `str` | 문자열 | `"절도"` | 죄명, 시설 이름 |
| `bool` | 참거짓 | `True`, `False` | 직업훈련 참여 여부 |

`type()`으로 값의 종류를 확인할 수 있습니다.

````python
나이 = 24
수용률 = 126.9
죄명 = "절도"
직업훈련 = False

print(type(나이))
print(type(수용률))
print(type(죄명))
print(type(직업훈련))
````

문자열은 따옴표로 감쌉니다. `24`는 숫자이고 `"24"`는 글자입니다. 둘은 다른 값입니다.

````python
print(24 == "24")
````

> [주의] True와 False는 첫 글자를 대문자로
>
> `true`, `false`로 쓰면 `NameError: name 'true' is not defined. Did you mean: 'True'?`가 납니다.


> [참고] 나누기 결과는 항상 실수
>
> `10 / 2`의 결과는 `5`가 아니라 `5.0`입니다. 정수 결과가 필요하면 몫 연산자 `//`를 씁니다.


> [주의] 실수 계산에는 아주 작은 오차가 있습니다
>
> ```python
> 0.1 + 0.2                    # 0.30000000000000004
> 0.1 + 0.2 == 0.3             # False
> round(0.1 + 0.2, 2) == 0.3   # True
> ```
>
> 컴퓨터가 실수를 2진수로 저장하기 때문에 생기는 현상입니다. 실수끼리 `==`로 비교할 때는 `round()`로 자릿수를 맞춘 뒤 비교합니다.

## 1.4 연산자

### 산술 연산자

| 연산자 | 뜻 | 예 | 결과 |
|---|---|---|---|
| `+` | 더하기 | `30 + 6` | `36` |
| `-` | 빼기 | `30 - 6` | `24` |
| `*` | 곱하기 | `30 * 2` | `60` |
| `/` | 나누기 | `30 / 12` | `2.5` |
| `//` | 몫 | `30 // 12` | `2` |
| `%` | 나머지 | `30 % 12` | `6` |
| `**` | 거듭제곱 | `2 ** 3` | `8` |

> [주의] 거듭제곱은 ^가 아니라 **
>
> 엑셀에서는 `2^3`이 8이지만 파이썬에서 `^`는 다른 연산(비트 연산)이라 `2 ^ 3`의 결과는 `1`입니다. 오류 없이 틀린 값이 나오므로 더 조심해야 합니다.


> [참고] 엑셀 수식과 다른 점
>
> | 엑셀 | 파이썬 |
> |---|---|
> | `=A1+B1` (등호로 시작) | `a + b` (등호 없이) |
> | `2^3` | `2 ** 3` |
> | `A1<>B1` | `a != b` |
> | `AND(조건1, 조건2)` | `조건1 and 조건2` |
> | `ROUND(x, 1)` | `round(x, 1)` (반올림 규칙이 다름, 아래 참고) |


몫(`//`)과 나머지(`%`)를 쓰면 형기 30개월을 "2년 6개월"로 바꿀 수 있습니다.

````python
형기_개월 = 30
년 = 형기_개월 // 12
개월 = 형기_개월 % 12
print(년, "년", 개월, "개월")
````

2025년 전국 교정시설의 수용 정원은 50,614명, 1일 평균 수용인원은 63,680명이었습니다(2026 교정통계연보). 수용률을 계산해 봅니다.

````python
정원 = 50614
현원 = 63680
수용률 = 현원 / 정원 * 100
print(수용률)
print(round(수용률, 1))
````

> [주의] 숫자에 쉼표를 넣지 않기
>
> ```python
> 정원 = 50,614
> print(정원)   # (50, 614)
> ```
>
> 쉼표를 넣으면 숫자 하나가 아니라 두 값의 묶음(튜플)이 됩니다. 오류가 나지 않아 뒤의 계산이 조용히 틀어집니다. 자릿수를 구분하고 싶으면 밑줄을 씁니다. `50_614`는 `50614`와 같습니다.


> [참고] round()의 반올림 규칙
>
> `round(2.5)`는 3이 아니라 `2`, `round(3.5)`는 `4`입니다. 정확히 가운데(.5)일 때 짝수 쪽으로 맞추기 때문입니다. 또 실수 오차 때문에 `round(0.125, 2)`는 `0.12`가 됩니다. 보고서 숫자가 엑셀과 끝자리 하나가 다르면 이 규칙을 먼저 의심합니다.


곱셈과 나눗셈이 덧셈과 뺄셈보다 먼저 계산됩니다. 먼저 계산할 부분은 괄호로 묶습니다.

````python
print(현원 - 정원)                 # 정원을 넘는 인원
print((현원 - 정원) / 정원 * 100)  # 정원 대비 초과 비율(%)
````

### 비교 연산자

비교 연산의 결과는 `True` 또는 `False`입니다.

| 연산자 | 뜻 |
|---|---|
| `==` | 같다 |
| `!=` | 다르다 |
| `>`, `>=` | 크다, 크거나 같다 |
| `<`, `<=` | 작다, 작거나 같다 |

````python
나이 = 24
죄명 = "절도"
print(나이 < 30)
print(나이 == 30)
print(죄명 == "사기")
print(죄명 != "사기")
````

`=`는 저장, `==`는 같은지 비교입니다. 처음에 가장 많이 헷갈리는 부분입니다.

> [팁] 범위는 한 번에 비교할 수 있습니다
>
> `20 <= 나이 < 30`처럼 쓰면 "20 이상 30 미만"을 한 번에 확인합니다. `나이 >= 20 and 나이 < 30`과 같습니다.


### 논리 연산자

| 연산자 | 뜻 |
|---|---|
| `and` | 둘 다 참이면 참 |
| `or` | 하나라도 참이면 참 |
| `not` | 참과 거짓을 뒤집음 |

````python
나이 = 24
징벌횟수 = 2
직업훈련 = False

print(나이 < 30 and 징벌횟수 > 0)
print(나이 < 30 or 징벌횟수 > 3)
print(not 직업훈련)
````

### 문자열 연산

문자열끼리 `+`로 이어 붙이고, `*`로 반복할 수 있습니다.

````python
시설 = "A" + "교도소"
print(시설)
print("-" * 20)
````

````python
print(f"{시설}의 수용률은 {round(수용률, 1)}%입니다")
````

> [주의] 문자열과 숫자는 바로 더할 수 없습니다
>

> [주의] 따옴표는 짝을 맞추기
>
> `print("안녕)`처럼 닫는 따옴표가 빠지면 `SyntaxError: unterminated string literal`이 납니다. f-문자열 안에서 딕셔너리 값을 꺼낼 때는 바깥과 다른 따옴표를 씁니다: `f"{수용자['나이']}세"`.

## 1.5 형변환

자료형을 바꿔야 할 때가 있습니다. 엑셀이나 CSV에서 읽은 숫자가 글자로 들어와 있는 경우가 대표적입니다.

````python
나이_문자 = "24"
나이_문자 + 1
````

위 칸은 일부러 오류가 나게 만든 코드입니다. 오류 메시지의 마지막 줄을 읽으면 원인을 알 수 있습니다. 여기서는 글자(`str`)와 숫자(`int`)를 더할 수 없다는 뜻입니다. `int()`로 숫자로 바꾸면 계산됩니다.

````python
print(int(나이_문자) + 1)
print(str(1001) + "번")
print(float("126.9"))
````

> [주의] 형변환에서 자주 나는 오류
>
> | 코드 | 결과 | 해결 |
> |---|---|---|
> | `int("24.5")` | `ValueError` | 소수점이 있는 글자는 `int(float("24.5"))` |
> | `float("1,234")` | `ValueError` | 쉼표를 지우고 변환: `float("1,234".replace(",", ""))` |
> | `int(24.9)` | `24` | 반올림이 아니라 소수점 아래를 버림. 반올림은 `round(24.9)` |

## 1.6 리스트

여러 값을 순서대로 묶을 때는 리스트를 씁니다. 대괄호 `[ ]` 안에 쉼표로 구분해 넣습니다.

````python
나이목록 = [24, 31, 45, 28, 52]
print(나이목록)
````

위치(인덱스)로 값을 꺼냅니다. 위치는 0부터 셉니다.

````python
print(나이목록[0])    # 첫 번째 값
print(나이목록[1])    # 두 번째 값
print(나이목록[-1])   # 마지막 값
print(나이목록[1:3])  # 두 번째부터 세 번째까지 (3번 위치는 포함하지 않음)
````

> [주의] 위치는 0부터, 끝 번호는 포함하지 않음
>
> - 첫 번째 값은 `[0]`입니다. 값이 5개인 리스트의 마지막은 `[4]` 또는 `[-1]`이고, `[5]`를 꺼내면 `IndexError: list index out of range`가 납니다.
> - `나이목록[1:3]`은 1번, 2번 위치만 꺼냅니다. 끝 번호(3)는 들어가지 않습니다.


리스트에 든 값의 개수는 `len()`으로 셉니다.

````python
print(len(나이목록))
````

`append()`로 맨 뒤에 값을 추가합니다.

````python
나이목록.append(37)
print(나이목록)
````

> [주의] append 결과를 다시 저장하지 않기
>
> ```python
> 나이목록 = 나이목록.append(37)
> print(나이목록)   # None
> ```
>
> `append()`는 리스트 자체를 바꾸고 아무것도 돌려주지 않습니다(`None`). 결과를 다시 저장하면 리스트가 사라집니다. `나이목록.append(37)`만 씁니다.


> [참고] 리스트를 다른 이름에 저장하면 같은 리스트를 가리킵니다
>
> ```python
> a = [1, 2]
> b = a
> b.append(3)
> print(a)   # [1, 2, 3]  (a도 바뀜)
> ```
>
> 따로 복사하려면 `b = a.copy()`를 씁니다.

## 1.8 반복문

리스트의 값을 하나씩 꺼내 같은 작업을 반복할 때 `for`를 씁니다. 형기를 하나씩 더해 합계를 구해 봅니다.

````python
형기목록 = [24, 36, 24, 30]
합계 = 0
for 형기 in 형기목록:
    합계 = 합계 + 형기
print("형기 합계:", 합계)
````

- `for` 줄 끝에 콜론(`:`)을 붙입니다.
- 반복할 줄은 4칸 들여 씁니다. 파이썬은 들여쓰기로 어디까지가 `for` 안인지 구분합니다.

아래 움직이는 그림은 위 코드가 한 줄씩 실행되는 과정입니다. 노란 줄이 방금 실행한 줄이고, 오른쪽에 그 결과로 바뀐 변수 값이 초록색으로 표시됩니다.

(그림: for 반복문이 한 줄씩 실행되는 과정 - 웹 강의자료 [https://statjeong.github.io/moj-python-lecture/01_python_basics.html](https://statjeong.github.io/moj-python-lecture/01_python_basics.html)에서 볼 수 있습니다)

반복할 때마다 `형기`와 `합계`가 바뀌는 과정을 표로 정리하면 다음과 같습니다.

| 반복 | 꺼낸 값 `형기` | 실행 전 `합계` | 실행 후 `합계` |
|---|---|---|---|
| 1회째 | 24 | 0 | 24 |
| 2회째 | 36 | 24 | 60 |
| 3회째 | 24 | 60 | 84 |
| 4회째 | 30 | 84 | 114 |

> [주의] 합계 변수는 반복문 위에서 0으로
>
> ```python
> for 형기 in 형기목록:
>     합계 = 0              # 반복할 때마다 0으로 돌아감
>     합계 = 합계 + 형기
> ```
>
> 이렇게 쓰면 마지막 값만 남습니다. 처음 값을 정하는 줄은 반복문 위에 둡니다.


> [참고] 들여쓰기가 반복 범위를 정합니다
>
> `for` 아래 들여 쓴 줄만 반복됩니다. `print("형기 합계:", 합계)`를 들여 쓰면 반복할 때마다 출력되고, 들여 쓰지 않으면 반복이 끝난 뒤 한 번만 출력됩니다.


> [주의] 반복문에서 자주 나는 오류
>
> | 실수 | 오류 메시지 |
> |---|---|
> | 콜론 빠뜨림 (`for 형기 in 형기목록`) | `SyntaxError: expected ':'` |
> | 반복할 줄을 들여 쓰지 않음 | `IndentationError: expected an indented block after 'for' statement on line 3` (끝의 숫자는 `for`가 있는 줄 번호) |
> | 같은 묶음 안에서 들여쓰기 칸 수가 다름 | `IndentationError: unindent does not match any outer indentation level` |


> [참고] 들여쓰기 칸 수는 묶음 안에서 같게
>
> Colab 편집기는 콜론 다음 줄을 자동으로 들여 씁니다. 설정에 따라 2칸 또는 4칸입니다. 칸 수 자체보다 같은 묶음 안에서 칸 수가 같은 것이 중요합니다. 다른 곳에서 복사해 온 코드와 섞으면 어긋나기 쉽습니다.

## 1.9 함수

자주 하는 계산을 함수로 만들어 두면, 값만 바꿔 넣어 같은 계산을 몇 번이고 다시 할 수 있습니다. 지금까지 쓴 `print()`, `len()`, `round()`, `type()`도 파이썬에 미리 만들어져 있는 함수입니다. 여기서는 함수를 직접 만듭니다.

### 함수 만들기와 부르기

````python
def 수용률계산(현원, 정원):
    비율 = 현원 / 정원 * 100
    return round(비율, 1)

A = 수용률계산(1523, 1200)
B = 수용률계산(1080, 900)
print("A교도소:", A, "/ B교도소:", B)
````

| 부분 | 뜻 |
|---|---|
| `def 수용률계산(현원, 정원):` | `수용률계산`이라는 함수를 만든다. 괄호 안은 함수에 넣을 값의 이름(매개변수) |
| 들여 쓴 줄 | 함수를 부를 때 실행되는 내용 |
| `return` | 결과를 돌려주고 함수를 끝낸다 |
| `수용률계산(1523, 1200)` | 함수를 부른다. 1523은 `현원`에, 1200은 `정원`에 들어간다 |

> [주의] print와 return은 다릅니다
>
> ```python
> def 수용률출력(현원, 정원):
>     print(현원 / 정원 * 100)
>
> 결과 = 수용률출력(1523, 1200)   # 화면에는 126.91...이 보이지만
> print(결과)                       # None
> ```
>
> `print()`는 화면에 보여주기만 하고 함수 밖으로 값을 돌려주지 않습니다. 결과를 변수에 저장해 다시 쓰려면 `return`이 있어야 합니다.


> [주의] 함수를 만든 칸을 먼저 실행하기
>
> 함수를 부르기 전에 `def`가 있는 칸을 먼저 실행해야 합니다. 그렇지 않으면 `NameError: name '수용률계산' is not defined`가 납니다. 함수 안쪽을 고쳤다면 `def` 칸을 다시 실행해야 반영됩니다.


아래 그림에서 함수를 부르면 값이 함수 안으로 들어가고(점선 상자), `return`으로 결과를 돌려준 뒤 함수 안의 변수가 사라지는 과정을 볼 수 있습니다. `def` 줄을 실행할 때는 함수를 만들어 두기만 하고, 안쪽 줄은 함수를 부를 때 실행됩니다.

(그림: 함수를 부르고 결과를 돌려받는 과정 - 웹 강의자료 [https://statjeong.github.io/moj-python-lecture/01_python_basics.html](https://statjeong.github.io/moj-python-lecture/01_python_basics.html)에서 볼 수 있습니다)

### 기본값과 이름 붙여 넣기

매개변수에 기본값을 정해 두면, 값을 넣지 않았을 때 기본값을 씁니다. 값을 넣을 때 `단위="개월"`처럼 이름을 붙여 넣을 수도 있습니다.

앞의 수용률 함수에 소수점 몇째 자리까지 남길지 정하는 `자리`를 더하고, 기본값을 1로 둡니다.

````python
def 수용률계산(현원, 정원, 자리=1):
    return round(현원 / 정원 * 100, 자리)

print(수용률계산(1523, 1200))            # 자리를 넣지 않으면 기본값 1
print(수용률계산(1523, 1200, 자리=2))    # 소수점 둘째 자리까지
print(수용률계산(현원=1080, 정원=900))   # 이름을 붙여 넣기
````

> [주의] 넣는 값의 개수와 순서
>
> - 매개변수가 두 개인 함수에 값을 하나만 넣으면 `TypeError: 수용률계산() missing 1 required positional argument: '정원'`이 납니다. 기본값이 있는 매개변수만 생략할 수 있습니다.
> - 이름을 붙인 값 뒤에 이름 없는 값을 쓰면 안 됩니다. `수용률계산(자리=2, 1523, 1200)`은 `SyntaxError: positional argument follows keyword argument`가 납니다. 이름 없는 값을 먼저 씁니다.
